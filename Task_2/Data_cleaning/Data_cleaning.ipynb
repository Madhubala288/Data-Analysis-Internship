## Handle Missing Values

```python
# Handle missing values in customer city  by replacing with 'Unknown'
df["Customer_City"] = df["Customer_City"].fillna("Unknown")
```

```python
# Handle missing values in payment method by replacing with 'Unknown'
df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")
```

```python
# Handle missing values in numerical column (customer age) by replacing with the median age
df["Customer_Age"] = df["Customer_Age"].fillna(df["Customer_Age"].median())
```

```python
df.isnull().sum()
```

### check duplicates

```python
df.duplicated().sum()
```

### drop duplicates

```python
df.drop_duplicates(
    inplace=True
)
```

```python
df.duplicated().sum()
```

## Standarized text columns

```python
# Standardize city names by converting to title case
df["Customer_City"] = df["Customer_City"].str.title()
```

```python
# Remove leading/trailing whitespaces and convert product categories to title case
df["Product_Category"] = df["Product_Category"].str.strip().str.title()
```

```python
# Remove leading/trailing whitespaces and convert payment methods to title case
df["Payment_Method"] = df["Payment_Method"].str.strip().str.title()
```

```python
# Verify the unique values in the Customer_City column
df["Customer_City"].unique()
```

### convert purchase dates into consistent datetime format

```python
df["Purchase_Date"] = pd.to_datetime(
    df["Purchase_Date"],
    errors="coerce"
)
```

```python
df.info()
```

```python
# Drop rows where Purchase_Date is missing (NaT)
df = df.dropna(subset=["Purchase_Date"])
```

```python
df.info()
```

## Handle Invalid Values

```python
# Check quantity column
df[df["Quantity"]<=0]
```

```python
# Remove 0 and negative values keep only possitive quantities
df=df[
    df["Quantity"]>0
]
```

```python
# check unit price column
df[df["Unit_Price"]<=0]
```

```python
#Remove 0 and negative values keep only possitive Unit prices
df=df[
    df["Unit_Price"]>0
]
```

```python
# check customer age column
df[df["Customer_Age"]<=0]
```

```python
# Remove unrelastic ages , keep only reliastic age betwee 10 and 100
df=df[
    (df["Customer_Age"]>=10)
    &
    (df["Customer_Age"]<=100)
]
```
